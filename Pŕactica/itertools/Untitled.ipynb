```python
l = [[random.randint(1,10) for i in range(4)] for j in range(4)]
keyfunc = lambda l: sum(l)/len(l)    
groups = []
keys = []
D = sorted(l, key=keyfunc)
for k, g in itertools.groupby(D, keyfunc):
    groups.append(list(g))
    keys.append(k)

print(list(zip(groups, keys)), sep='\n')
```

```python
from typing import List

def P(l: List[int], f=print):
    if len(l) == 1:
        return l
    for i in range(len(l)):
        copy = l.copy()
        removed_item = copy.pop(i)  # Get the actual item value
        
        # Recurse and capture the returned sub-list
        sub_permutation = P(copy, f)
        
        # Combine them into a full permutation list
        full_path = [removed_item] + sub_permutation
        
        # Only print at the top-level frame to avoid printing partial snippets
        if len(l) == 4: 
            f(full_path)
            
    return l # Return a valid list to prevent NoneType errors upstream

P([1,2,3,4])
```
